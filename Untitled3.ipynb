```python
import pandas as pd
import plotly.graph_objects as go

df_velas = df_ecopetrol.iloc[1:].copy()

df_velas['Fecha'] = pd.to_datetime(df_velas['Fecha'], errors='coerce')

# Limpiamos y convertimos las columnas reales a numérico
df_velas['cierre'] = pd.to_numeric(df_velas['Precio cierre'].astype(str).str.replace(',', ''), errors='coerce')
df_velas['maximo'] = pd.to_numeric(df_velas['Precio máximo'].astype(str).str.replace(',', ''), errors='coerce')
df_velas['minimo'] = pd.to_numeric(df_velas['Precio mínimo'].astype(str).str.replace(',', ''), errors='coerce')
# Usamos el cierre como aproximación de apertura al no tener esa columna explícita
df_velas['apertura'] = df_velas['cierre']

df_velas = df_velas.dropna(subset=['Fecha', 'apertura', 'cierre']).sort_values('Fecha')

fig = go.Figure(data=[go.Candlestick(x=df_velas['Fecha'],
                open=df_velas['apertura'],
                high=df_velas['maximo'],
                low=df_velas['minimo'],
                close=df_velas['cierre'],
                increasing_line_color='green', decreasing_line_color='red')
])

fig.update_layout(
    title='Gráfico de Velas Japonesas - ECOPETROL',
    yaxis_title='Precio (COP)',
    xaxis_title='Fecha',
    xaxis_rangeslider_visible=True
)

fig.show()
```

```python
import pandas as pd
from google.colab import files


df_limpio = df_ecopetrol.copy()


df_limpio['Fecha'] = pd.to_datetime(df_limpio['Fecha'], errors='coerce')


columnas_numericas = ['Precio cierre', 'Precio máximo', 'Precio promedio ponderado', 'Precio mínimo', 'Cantidad', 'Volumen']

for col in columnas_numericas:
    if col in df_limpio.columns:
        df_limpio[col] = pd.to_numeric(df_limpio[col].astype(str).str.replace(',', ''), errors='coerce')

output_filename = 'ECOPETROL_limpio.csv'
df_limpio.to_csv(output_filename, sep=';', index=False, encoding='utf-8')

print("Archivo procesado correctamente. Iniciando descarga...")


files.download(output_filename)
```
